# Evaluation des données
## Import des données

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
loyers_df = pd.read_csv('./data/loyers.csv', sep=';', encoding='latin-1', decimal=',')
dvf_df = pd.read_csv('./data/dvf.csv', sep=';', encoding='latin-1', decimal=',')
```

## Préparation des données

### Nettoyage des données des loyers

```python
loyers_df = loyers_df[['INSEE_C', 'LIBGEO', 'annee', 'loypredm2', 'lwr.IPm2', 'upr.IPm2']].sort_values(['INSEE_C', 'annee'])
```

### Création du code commune pour les DVF
Concaténation du `Code departement` et du `Code commune`.

```python
dvf_df['INSEE_C'] = (dvf_df['Code departement'].astype(str).str.zfill(2) + dvf_df['Code commune'].astype(str).str.zfill(3)).astype('category')
dvf_df.head(5)
```

```python
loyers_df['INSEE_C'] = loyers_df['INSEE_C'].astype('category')
loyers_df.head(5)
```

### Calcul des DVF au m²

```python
dvf_df['VFm2'] = dvf_df['Valeur fonciere'] / dvf_df['Surface Carrez du 1er lot']

dvf_df
```

### Aggergation des DVF par année

#### Etude de la distribution des prix de vente par departement

```python
dvf_df[['Nature mutation', 'Type local', 'Nombre pieces principales']].value_counts()
```

```python
dvf_df['Code departement'] = dvf_df['Code departement'].astype('category')

fig = px.box(dvf_df, x='Code departement', y='VFm2', color='Nombre pieces principales', title="Distribution des valeurs foncières (VF) au m² pour un appartement de 2 pièces maximum")
fig.show()
```

Il y a trop de volatilité sur les données de ventes. 

Nous allons utiliser la méthode de l'interquartile (**IQR**) pour nous séparer des valeurs aberrantes.

```python
# Calculer Q1, Q3 et IQR
Q1 = dvf_df['VFm2'].quantile(0.25)
Q3 = dvf_df['VFm2'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filtrer les données
dvf_df_clean = dvf_df[(dvf_df['VFm2'] >= lower_bound) & (dvf_df['VFm2'] <= upper_bound)]
print(f"Données supprimées: {len(dvf_df) - len(dvf_df_clean)} ({100*(len(dvf_df) - len(dvf_df_clean))/len(dvf_df):.1f}%)")
```

```python
fig = px.box(dvf_df_clean, x='Code departement', y='VFm2', color='Nombre pieces principales', title="Distribution des valeurs foncières (VF) au m² pour un appartement de 2 pièces maximum (Sans valeurs aberrantes)")
fig.show()
```

Nous allons aggréger les `Valeur fonciere` par année et par commune en utilsant le prix median car il y a toujours **trop de volatilité sur les ventes**.

#### Aggregation

```python
agg_dvf_df = dvf_df_clean[['INSEE_C', 'annee', 'VFm2']]\
                        .groupby(['INSEE_C', 'annee'])\
                        .agg({'VFm2': np.median})\
                        .reset_index()\
                        .sort_values(by=['INSEE_C', 'annee'])

agg_dvf_df.columns = ['INSEE_C', 'annee', 'VFm2med']
```

Remplissage des valeurs nulles en utilisant la **moyenne** de `VFmed` par `INSEE_C`.

```python
# Calcul des valeurs de ventes moyennes par commune
means = agg_dvf_df.groupby('INSEE_C')['VFm2med'].mean().to_dict()
# Application des valeurs de ventes moyennes sur les valeurs nulles de la commune concernée
agg_dvf_df['VFm2med'] = agg_dvf_df['VFm2med'].fillna(agg_dvf_df['INSEE_C'].map(means))

agg_dvf_df
```

## Jointure des loyers avec les DVF sur le code commune/année

On supprime également les communes pour lesquelles nous n'avons pas de ventes.

```python
full_df = loyers_df.merge(agg_dvf_df, how='left', on=['INSEE_C', 'annee']).dropna().sort_values(['INSEE_C', 'annee'])

full_df
```

## Calcul du taux de rentabilité 

La formule utilisée est : $((`loypredm2` * 12) / `VFm2med`) * 100$

(Suppression des taux de rentabilité aberrants *> 15%*)

```python
full_df['txRenta'] = ((full_df['loypredm2'] * 12) / full_df['VFm2med']) * 100

full_df = full_df[full_df['txRenta'] < 15]
full_df
```
